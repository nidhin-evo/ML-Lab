```python
import pandas as pd

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB, BernoulliNB
from sklearn.metrics import accuracy_score, f1_score
```

```python
train_data = fetch_20newsgroups(
    subset="train",
    remove=("headers", "footers", "quotes")
)

test_data = fetch_20newsgroups(
    subset="test",
    remove=("headers", "footers", "quotes")
)

X_train = train_data.data
y_train = train_data.target

X_test = test_data.data
y_test = test_data.target

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))
```

```python
vectorizer_m = CountVectorizer(
    stop_words="english",
    min_df=2
)

X_train_m = vectorizer_m.fit_transform(X_train)
X_test_m = vectorizer_m.transform(X_test)

print("Training matrix:", X_train_m.shape)
print("Testing matrix :", X_test_m.shape)
```

```python
mnb = MultinomialNB()

mnb.fit(X_train_m, y_train)

y_pred_mnb = mnb.predict(X_test_m)

print("Multinomial Naive Bayes trained successfully!")
```

```python
vectorizer_b = CountVectorizer(
    binary=True,
    stop_words="english",
    min_df=2
)

X_train_b = vectorizer_b.fit_transform(X_train)
X_test_b = vectorizer_b.transform(X_test)

bnb = BernoulliNB()

bnb.fit(X_train_b, y_train)

y_pred_bnb = bnb.predict(X_test_b)

print("Bernoulli Naive Bayes trained successfully!")
```

```python
mnb_accuracy = accuracy_score(y_test, y_pred_mnb)
mnb_f1 = f1_score(y_test, y_pred_mnb, average="weighted")

bnb_accuracy = accuracy_score(y_test, y_pred_bnb)
bnb_f1 = f1_score(y_test, y_pred_bnb, average="weighted")

print("Multinomial NB")
print("Accuracy:", round(mnb_accuracy, 4))
print("F1 Score:", round(mnb_f1, 4))

print("\nBernoulli NB")
print("Accuracy:", round(bnb_accuracy, 4))
print("F1 Score:", round(bnb_f1, 4))
```

```python
results = pd.DataFrame({
    "Model": ["Multinomial NB", "Bernoulli NB"],
    "Accuracy": [mnb_accuracy, bnb_accuracy],
    "F1 Score": [mnb_f1, bnb_f1]
})

print(results.to_string(index=False))
```
